```python
import cv2 # Load OpenCV image functions.
import numpy as np # Load NumPy as np for array operations.
import matplotlib.pyplot as plt # Load plotting functions as plt.
filename = "images/mandrill.tiff" # Select the image file to observe.
bgr = cv2.imread(filename) # Read the file in BGR channel order.
if bgr is None: # Check whether the file could not be read.
 raise FileNotFoundError(filename) # Stop and show the file name.
rgb = cv2.cvtColor( # Make a color image for Matplotlib display.
 bgr, cv2.COLOR_BGR2RGB) # Reorder BGR channels into RGB.
gray = cv2.cvtColor( # Make a one-channel grayscale image.
 bgr, cv2.COLOR_BGR2GRAY) # Convert color to intensity.
print("Shape:", gray.shape, # Print height and width.
 "Type:", gray.dtype) 
```

```python
hist = cv2.calcHist( # Count pixels at each grayscale intensity.
 [gray], [0], None, # Use channel 0 and include every pixel.
 [256], [0, 256]) # Use 256 bins and include values 0 to 255.
fig, axes = plt.subplots( # Make a figure with three panels.
 1, 3, figsize=(14, 4)) # One row, three columns, size in inches.
axes[0].imshow(rgb) # Show the RGB image in the first panel.
axes[0].set_title(filename + " RGB") # Name the first panel.
axes[0].axis("off") # Hide the image coordinate axes.
axes[1].imshow( # Show grayscale in the second panel.
 gray, cmap="gray", # Use the black-to-white color map.
 vmin=0, vmax=255) # Fix black at 0 and white at 255.
axes[1].set_title("Grayscale") # Name the second panel.
axes[1].axis("off") # Hide the image coordinate axes.
axes[2].plot( # Draw the histogram in the third panel.
 np.arange(256), # Set x values to integers 0 through 255.
 hist.ravel()) # Flatten the bin counts for the y values.
axes[2].set_xlim(0, 255) # Fix the intensity axis limits.
axes[2].set_xlabel("Intensity value") # Label the horizontal axis.
axes[2].set_ylabel("Pixel count") # Label the vertical axis.
plt.tight_layout() # Adjust spacing to reduce overlap.
plt.show() 
```

```python
print("Pixels:", gray.size) # Print the total number of pixels.
print("Histogram total:", # Name the total count in the output.
 int(hist.sum())) # Sum all bins and print an integer.
print("Minimum:", int(gray.min())) # Lowest intensity present.
print("Maximum:", int(gray.max())) # Highest intensity present.
print("Mean:", # Name the average intensity in the output.
 round(float(gray.mean()), 1)) # Round the mean to one decimal.
print("Std:", # Name the global intensity spread in the output.
 round(float(gray.std()), 1)) # Round the spread to one decimal.
peak = int(np.argmax(hist)) # Find the most frequent intensity.
print("Most frequent intensity:", peak) # Print that intensity.
print("Pixels at that intensity:", # Name the peak bin count.
 int(hist.ravel()[peak])) 
```

```python
colors = ("red", "green", "blue") # Match RGB channel order.
plt.figure(figsize=(8, 4)) # Make one figure, 8 by 4 inches.
for channel, color in enumerate(colors): # Pair index and color.
 channel_hist = cv2.calcHist( # Count one color channel.
 [rgb], [channel], None, # Select the channel, all pixels.
 [256], [0, 256]) # Use 256 bins for values 0 to 255.
 plt.plot(channel_hist.ravel(), # Plot counts at indices 0 to 255.
 color=color, # Draw this line in its channel color.
 label=color.upper()) # Name the line RED, GREEN or BLUE.
plt.xlim(0, 255) # Fix the channel intensity axis limits.
plt.xlabel("Channel intensity") # Label the horizontal axis.
plt.ylabel("Pixel count per channel") # Label the vertical axis.
plt.title(filename + " RGB histogram") # Add the image file name.
plt.legend() # Show the label for each colored line.
plt.tight_layout() # Adjust plot spacing to fit the labels.
plt.show() 
```
